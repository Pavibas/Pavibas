```python
db={'pavibas05' : 'Password@123'}
username=input('Enter the userame')
password=input('Enter the password')

def loginerror(username,password):
    if username in db.keys() and password==db[username]:
        return True
    else:
        raise NameError('Invalide username and password')
def login(username, password):
    try:
        df=loginerror(username,password)
        if df==True:
            print('login is successfull')
    except Exception as e:
        print('login is unsuccessfull -', e)
    else:
        print('Dear students, Welcome')

if __name__=='__main__':
    login(username, password)
    
```
